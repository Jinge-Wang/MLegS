```python
def calculate_total_disk_usage(disk_usage_text):
    """
    Calculate total disk usage from a text output similar to 'du -h' command.
    
    Args:
        disk_usage_text (str): Text containing disk usage information
    
    Returns:
        float: Total disk usage in GB
    """
    lines = disk_usage_text.strip().split('\n')
    total_bytes = 0
    
    for line in lines:
        # Split by whitespace (not tabs)
        parts = line.split(maxsplit=1)
        if len(parts) >= 1:
            size_str = parts[0].strip()
            
            # Handle size formats like "762M", "1.9G", etc.
            if size_str[-1] in ['K', 'M', 'G', 'T']:
                unit = size_str[-1]
                try:
                    number = float(size_str[:-1])
                    
                    # Convert to bytes
                    if unit == 'K':
                        bytes_val = number * 1024
                    elif unit == 'M':
                        bytes_val = number * 1024 * 1024
                    elif unit == 'G':
                        bytes_val = number * 1024 * 1024 * 1024
                    elif unit == 'T':
                        bytes_val = number * 1024 * 1024 * 1024 * 1024
                    else:
                        bytes_val = number
                    
                    total_bytes += bytes_val
                except ValueError:
                    print(f"Could not parse size from: {size_str}")
    
    # Convert total bytes to GB
    total_gb = total_bytes / (1024 * 1024 * 1024)
    return total_gb

# Obtain the disk list from file
with open('disk_list.txt', 'r') as file:
    disk_usage_data = file.read()

total_usage = calculate_total_disk_usage(disk_usage_data)
print(f"Total disk usage: {total_usage:.2f} GB")
```
